```python
import pandas as pd
```

```python
df_final = pd.read_excel("C:/pessoal/projetos_individuais/ia/Iso_olhos/dados/Pacientes_Torica_FINAL.xlsx")
```

```python
print(df_final.columns)
```

```python
cols = ['k1', 'eixo_plano', 'k2',
       'eixo_curvo', 'al', 'acd', 'lt', 'wtw']

# Nome da coluna que identifica a linha (ex: primeira coluna do df)
id_col = df_final.columns[0]

# Armazenar resultados
non_numeric_rows = []

for col in cols:
    mask = pd.to_numeric(df_final[col], errors='coerce').isna() & df_final[col].notna()
    for idx in df_final[mask].index:
        non_numeric_rows.append({
            'id': df_final.loc[idx, id_col],
            'column': col,
            'value': df_final.loc[idx, col]
        })

# Transformando em DataFrame para visualizar melhor
non_numeric_df = pd.DataFrame(non_numeric_rows)
print(non_numeric_df)
```

```python
cols_to_replace = ['k1', 'eixo_plano', 'k2', 'eixo_curvo', 'al', 'acd', 'lt', 'wtw']

# Substituindo . por , em todas as colunas
df_final[cols_to_replace] = df_final[cols_to_replace].astype(str).apply(lambda x: x.str.replace('.', ',', regex=False))

# Conferir resultado
print(df_final[cols_to_replace].head())
```

```python
df_final.to_excel("C:/pessoal/projetos_individuais/ia/Iso_olhos/dados/Pacientes_Torica_FINAL_check.xlsx", index=False)
```

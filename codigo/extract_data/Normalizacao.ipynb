```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_excel("C:/pessoal/projetos_individuais/ia/Iso_olhos/dados/Pacientes_Torica_FINAL_atualizados.xlsx")
```

```python
print (df.columns)
```

```python
from sklearn.impute import SimpleImputer
# 1️⃣ Colunas a manter sem normalizar
cols_excluir = [
    'pront', 'sexo', 'olho_operado'
]

df = df[~df['topografia'].isin(['T']) & df['topografia'].notna()]
df['topografia'] = df['topografia'].str.extract(r'T(\d+)').astype(float)
print(df['topografia'].unique())

# 2️⃣ Detectar colunas one-hot (começam com "tipo_")
cols_one_hot = [c for c in df.columns if c.startswith("tipo_")]

# 3️⃣ Colunas que serão normalizadas
cols_normalizar = [
    c for c in df.columns 
    if c not in cols_excluir + cols_one_hot
]
cols_num = [c for c in df.columns if c not in cols_excluir + cols_one_hot]
imputer = SimpleImputer(strategy='mean')
df[cols_num] = imputer.fit_transform(df[cols_num])

# 4️⃣ Criar scaler e aplicar
scaler = StandardScaler()
df[cols_normalizar] = scaler.fit_transform(df[cols_normalizar])
```

```python
df.to_excel("C:/pessoal/projetos_individuais/ia/Iso_olhos/dados/Pacientes_Torica_FINAL_check.xlsx", index=False)
```
